## Part 1

```python
random.seed(42)
np.random.seed(42)
# # random points
# n = 5
# points = np.random.rand(n, 2)

# Punkter på en cirkel
n = 200
theta = np.linspace(0, 2*np.pi, n, endpoint=False)

points = np.column_stack((np.cos(theta), np.sin(theta)))
```

```python
random.seed(42)
np.random.seed(42)

# cost matrix euclidean
def compute_cost_matrix(points):
    n = len(points)
    A = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            A[i, j] = np.linalg.norm(points[i] - points[j])
    return A

# cost function
def route_cost(route, cost_matrix):
    return sum(cost_matrix[route[i], route[(i+1) % len(route)]]
               for i in range(len(route)))

# swap cities, (1,2,3,4,5) -> (1,2,5,4,3)
def propose(route):
    new_route = route.copy()
    i, j = np.random.choice(len(route), size=2, replace=False)
    new_route[i], new_route[j] = new_route[j], new_route[i]
    return new_route

# cooling schedule  
def tk1(k):
    return 1 / np.sqrt(1 + k)

def tk2(k):
    return 1 / np.log(1 + k)
```

```python
random.seed(42)
np.random.seed(42)

cost = compute_cost_matrix(points)

K = 20000 # antal iteratiosn
route = np.arange(n)
np.random.shuffle(route)
current_cost = route_cost(route, cost)

for k in range(K):
    T = tk1(k)
    candidate = propose(route)
    candidate_cost = route_cost(candidate, cost)

    delta = candidate_cost - current_cost

    if delta < 0 or np.random.rand() < np.exp(-delta / T):
        route = candidate
        current_cost = candidate_cost

plt.figure(figsize=(6, 6))
plt.scatter(points[:, 0], points[:, 1], c='black')

for i in range(n):
    p1 = points[route[i]]
    p2 = points[route[(i+1) % n]]
    plt.plot([p1[0], p2[0]], [p1[1], p2[1]], 'b-')

plt.title(f"Simulated Annealing TSP — final cost = {current_cost:.2f}")
plt.show()
```

## Part 2

```python
random.seed(42)
np.random.seed(42)

# general cost matrix
def compute_cost_matrix(points, cost_function):
    n = len(points)
    A = np.zeros((n, n))
    for i in range(n):
        for j in range(n):
            A[i, j] = cost_function(points[i], points[j])
    return A

def euclidean_cost(p, q):
    return np.linalg.norm(p - q)

def manhattan_cost(p, q):
    return np.sum(np.abs(p - q))
```

```python
random.seed(42)
np.random.seed(42)

cost = compute_cost_matrix(points, manhattan_cost)

K = 20000 # antal iteratiosn
route = np.arange(n)
np.random.shuffle(route)
current_cost = route_cost(route, cost)

for k in range(K):
    T = tk1(k)
    candidate = propose(route)
    candidate_cost = route_cost(candidate, cost)

    delta = candidate_cost - current_cost

    if delta < 0 or np.random.rand() < np.exp(-delta / T):
        route = candidate
        current_cost = candidate_cost

plt.figure(figsize=(6, 6))
plt.scatter(points[:, 0], points[:, 1], c='black')

for i in range(n):
    p1 = points[route[i]]
    p2 = points[route[(i+1) % n]]
    plt.plot([p1[0], p2[0]], [p1[1], p2[1]], 'b-')

plt.title(f"Simulated Annealing TSP — final cost = {current_cost:.2f}")
plt.show()
```
